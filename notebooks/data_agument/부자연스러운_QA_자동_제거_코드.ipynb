```python
import json

# OpenAI API 클라이언트 (사전 정의되어 있다면 이 부분 생략)
from openai import OpenAI
client = OpenAI(api_key=MY_KEY)  # 환경변수 사용 시 생략 가능
```

```python
def filter_unnatural_qa_batch(qa_batch, title="", context=""):
    """
    QA 리스트 중 부자연스러운 질문-답변 쌍을 제거.
    question-answer 관계가 어색하거나, answer가 title 또는 context 내용과 맞지 않는 항목을 필터링.
    """
    qa_json = json.dumps(qa_batch, ensure_ascii=False, indent=2)
    prompt = f"""
당신은 한국어 Extractive QA 데이터셋 검수 전문가입니다.

아래는 자동으로 생성된 QA 데이터입니다.
각 항목의 question과 answer가 자연스럽고 실제로 context의 일부 내용을 기반으로 한 경우만 남기십시오.

제거 조건:
- question과 answer의 관계가 부자연스럽거나, answer가 질문의 답이 아닌 경우
- answer가 context에 포함되지 않는 경우
- answer가 title과 동일하거나 title 일부일 경우
- question이나 answer가 너무 짧거나 불완전할 경우

남길 QA만 JSON 배열로 출력하고, 그 외의 설명, 문장, 코멘트는 절대 포함하지 마십시오.

title: {title}
context: {context}

QA 데이터:
{qa_json}
"""

    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": "당신은 QA 데이터 검수 전문가입니다."},
                {"role": "user", "content": prompt}
            ],
            temperature=0.2
        )
        output = response.choices[0].message.content.strip()
        return json.loads(output)
    except Exception as e:
        print("검수 중 오류:", e)
        return qa_batch  # 실패 시 원본 반환
```

```python
# === 실행 예시 ===

input_path = "/content/augmented_dataset2.json"
output_path = "/content/drive/MyDrive/Colab Notebooks/odqa_data/data/generated_qa_filtered.json"

# 1️⃣ 기존 QA 로드
with open(input_path, "r", encoding="utf-8") as f:
    all_qas = json.load(f)

print(f"총 QA 문항 수: {len(all_qas)}")

# 2️⃣ 배치 단위로 필터링 (예: 20개씩)
filtered_results = []
batch_size = 20

for i in range(0, len(all_qas), batch_size):
    batch = all_qas[i:i+batch_size]
    title = batch[0].get("title", "") if batch else ""
    context = batch[0].get("context", "") if batch else ""

    refined_batch = filter_unnatural_qa_batch(batch, title, context)
    filtered_results.extend(refined_batch)

    print(f"처리 중... {i+len(batch)} / {len(all_qas)} 완료")

# 3️⃣ 결과 저장
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(filtered_results, f, ensure_ascii=False, indent=2)

print(f"필터링 완료. 최종 QA 문항 수: {len(filtered_results)}")
print(f"저장 위치: {output_path}")
```
